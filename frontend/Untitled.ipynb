```python
# ===== HomeMatch AI: Fix layout, synchronisation and donut chart =====

# 1. Make the numeric input boxes wider
for k in names:
    numbers[k].layout.width = "160px"
    numbers[k].layout.min_width = "160px"
    numbers[k].layout.max_width = "160px"
    numbers[k].layout.flex = "0 0 160px"

    sliders[k].layout.width = "auto"
    sliders[k].layout.min_width = "200px"
    sliders[k].layout.flex = "1 1 auto"

# Ensure each preference row uses the available width
for row in preference_rows:
    row.layout.width = "100%"
    row.layout.align_items = "center"
    row.layout.gap = "12px"

# 2. Remove old callbacks to prevent conflicting updates
for k in names:
    sliders[k].unobserve_all()
    numbers[k].unobserve_all()

updating = False

# 3. Rebalance the other five weights using integer percentages
def rebalance_weights(changed_key, new_value):
    global updating

    if updating:
        return

    keys = list(names.keys())
    others = [k for k in keys if k != changed_key]

    selected = max(0, min(100, int(new_value)))
    remaining = 100 - selected

    old = {k: int(numbers[k].value) for k in others}
    old_total = sum(old.values())

    if old_total:
        exact = {
            k: remaining * old[k] / old_total
            for k in others
        }
    else:
        exact = {k: remaining / len(others) for k in others}

    allocated = {k: int(exact[k]) for k in others}
    leftover = remaining - sum(allocated.values())

    order = sorted(
        others,
        key=lambda k: exact[k] - allocated[k],
        reverse=True
    )

    for k in order[:leftover]:
        allocated[k] += 1

    updating = True
    try:
        # Update the selected preference and all other preferences
        numbers[changed_key].value = selected
        sliders[changed_key].value = selected

        for k in others:
            numbers[k].value = allocated[k]
            sliders[k].value = allocated[k]
    finally:
        updating = False

    # 4. Refresh the donut chart and total label
    update_chart()


# 5. Both sliders and numeric boxes use the same update function
def slider_changed(change, key):
    if change["name"] == "value" and not updating:
        rebalance_weights(key, change["new"])


def number_changed(change, key):
    if change["name"] == "value" and not updating:
        rebalance_weights(key, change["new"])


for k in names:
    sliders[k].observe(
        lambda change, key=k: slider_changed(change, key),
        names="value"
    )
    numbers[k].observe(
        lambda change, key=k: number_changed(change, key),
        names="value"
    )

# 6. Refresh the chart immediately
update_chart()

print("Fix applied. Six preference weights should total 100%.")
```
